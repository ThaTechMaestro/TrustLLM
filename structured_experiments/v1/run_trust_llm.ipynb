```python
from trustllm.generation.generation import LLMGeneration

test_categories = ["fairness", "safety", "truthfulness", "ethics"]

'''
Using a concise dataset via dataset/dataset_mini
Larger dataset can be found via dataset/dataset
'''
for tc in test_categories:
    print(f"\n--- Running TrustLLM evaluation for category: {tc} ---")

    llm_gen = LLMGeneration(
        model_path="distilgpt2",
        test_type=tc,
        data_path="dataset/dataset_mini",
        online_model=True,
        max_new_tokens=256,
        debug=True
    )


    llm_gen.generation_results()
```
